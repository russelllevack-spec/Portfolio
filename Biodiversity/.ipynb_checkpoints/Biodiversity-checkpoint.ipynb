```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

species = pd.read_csv('species_info.csv')
observations = pd.read_csv('observations.csv')
total = pd.merge(observations, species, on='scientific_name', how='inner')
endangered = total[total.conservation_status == 'Endangered']
print(endangered.groupby(['scientific_name', 'park_name']).head())
parks = endangered.park_name.unique()
print(parks)
for i in parks:
    data = endangered[endangered.park_name == i]
    plt.figure(figsize=(10, 8))
    sns.barplot(data=data, x='scientific_name', y='observations')
    plt.xticks(rotation=45)
    plt.show()
    plt.close()
```
